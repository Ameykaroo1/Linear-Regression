```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("Salary Data.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df = df.fillna(df.mean(numeric_only=True))
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Education Level'] = df['Education Level'].fillna(df['Education Level'].mode()[0])
df['Job Title'] = df['Job Title'].fillna(df['Job Title'].mode()[0])
```

```python
df.isnull().sum()
```

```python
X=df.drop(columns=['Education Level','Salary'])
y=df['Salary']
X['Gender'] = X['Gender'].map({'Male':1,'Female':0})
X=pd.get_dummies(X,columns=['Job Title'],dtype=int)
```

```python
X.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.3,random_state=42)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import r2_score
```

```python
r2 =r2_score(y_test,y_pred)
print(r2)
n=X_test.shape[0]
p=X_test.shape[1]

adjusted_r2 = 1-((1-r2)*(n-1)/(n-p-1))
print(adjusted_r2)
```
